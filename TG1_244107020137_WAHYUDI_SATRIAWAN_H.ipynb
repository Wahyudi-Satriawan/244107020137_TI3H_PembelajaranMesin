### Fungsi dan Kegunaan Library

1. **PyPREP (Python Preprocessing Pipeline)**
   - **Fungsi:** Prapemrosesan otomatis untuk sinyal elektroensefalogram (EEG / rekaman aktivitas otak).
   - **Fitur Utama:** Mendeteksi kanal elektroda yang rusak (*bad channels*), membersihkan gangguan frekuensi listrik (*line noise*), dan melakukan standardisasi referensi sinyal (*robust referencing*).
   - **Kegunaan:** Menyiapkan data sinyal otak mentah agar bersih dari derau biologis maupun teknis sebelum dimasukkan ke model klasifikasi (misalnya untuk sistem *Brain-Computer Interface*).

2. **SciPy**
   - **Fungsi:** Library dasar komputasi saintifik, matematika, dan manipulasi data numerik tingkat lanjut berbasis NumPy.
   - **Fitur Utama:** Menyediakan modul optimasi fungsi (`scipy.optimize`), pengolahan sinyal digital (`scipy.signal`), aljabar linier lanjut (`scipy.linalg`), dan uji distribusi statistik (`scipy.stats`).
   - **Kegunaan:** Menghitung operasi aljabar pada data, ekstraksi fitur gelombang/citra, serta menjadi fondasi algoritma *machine learning* (seperti yang digunakan internal pada Scikit-Learn).

3. **wandb (Weights & Biases)**
   - **Fungsi:** Platform MLOps untuk pelacakan eksperimen (*experiment tracking*) dan pencatatan metrik performa model.
   - **Fitur Utama:** Mencatat grafik *loss* dan akurasi secara *real-time*, menyimpan konfigurasi hyperparameter, membandingkan performa beberapa arsitektur model, dan versioning dataset/model.
   - **Kegunaan:** Membantu dokumentasi pelatihan model agar riwayat eksperimen transparan, mudah dianalisis, dan dapat direproduksi (*reproducible*).

4. **pyECG**
   - **Fungsi:** Library analisis dan ekstraksi fitur khusus untuk sinyal elektrokardiogram (EKG / rekaman ritme jantung).
   - **Fitur Utama:** Menghilangkan derau dasar (*baseline wander*), mendeteksi puncak kompleks QRS (gelombang R), dan menghitung variabilitas detak jantung (*Heart Rate Variability* / HRV).
   - **Kegunaan:** Menyiapkan fitur statistik dan interval sinyal jantung untuk klasifikasi kondisi medis, seperti deteksi aritmia.

### 1. Pelanggaran Etika dan Hukum dalam Pemanfaatan AI

#### A. Pelanggaran Hak Cipta dan Eksploitasi Data Latih (*The New York Times v. OpenAI & Microsoft*)
- **Kasus:** Sengketa hukum lanjutan sepanjang 2024–2025 terkait gugatan The New York Times terhadap OpenAI dan Microsoft. Penggugat mempermasalahkan praktik *web scraping* jutaan artikel jurnalistik berhak cipta secara masif untuk melatih model GPT tanpa izin lisensi, royalti, maupun atribusi.
- **Aspek Hukum:** Dugaan pelanggaran hak kekayaan intelektual (*Copyright Act* AS), penyalahgunaan doktrin penggunaan wajar (*fair use*), serta ketidakpatuhan terhadap kewajiban transparansi data latih yang kini diatur ketat dalam kerangka hukum seperti *EU AI Act*.
- **Aspek Etika:** Pemanfaatan karya komersial pihak lain tanpa kompensasi yang adil, penurunan nilai ekonomi bagi institusi media orisinal, serta ketiadaan audit asal-usul data latih (*provenance transparency*).

#### B. Rekayasa Identitas Digital Menggunakan *Deepfake* untuk Kejahatan Finansial
- **Kasus:** Insiden penipuan finansial berskala besar di Hong Kong pada awal tahun 2024, di mana seorang staf keuangan mentransfer HK$200 juta (sekitar US$25,6 juta) setelah diperdaya dalam panggilan video konferensi. Seluruh peserta rapat dan pimpinan (CFO) dalam panggilan tersebut merupakan hasil manipulasi audio-visual sintetis (*multi-person deepfake*).
- **Aspek Hukum:** Tindak pidana penipuan terorganisir (*wire fraud*), pemalsuan identitas digital, dan pelanggaran hukum siber.
- **Aspek Etika:** Manipulasi kebenaran faktual secara visual/audio, pelanggaran hak integritas identitas individu, serta tidak adanya penanda digital (*watermarking* / C2PA metadata) wajib pada perangkat lunak generatif yang disalahgunakan.

### 2. Dampak Energi dan Lingkungan terhadap AI

#### A. Dampak Nyata terhadap Lingkungan
1. **Lonjakan Konsumsi Listrik dan Beban Grid:**
   Pusat data kecerdasan buatan membutuhkan daya komputasi kluster akselerator (GPU/TPU) yang berjalan nonstop. Berdasarkan laporan *Electricity 2024* dari International Energy Agency (IEA), konsumsi listrik global untuk pusat data, AI, dan sektor kripto diproyeksikan melonjak dua kali lipat menjadi lebih dari 1.000 TWh pada periode 2024–2026—setara dengan total konsumsi listrik satu negara Jepang. Emisi karbon tidak langsung (*Scope 2*) meningkat pesat terutama di kawasan pusat data yang masih bergantung pada pembangkit listrik berbasis bahan bakar fosil.
2. **Krisis Air Bersih untuk Pendinginan (*Water Footprint*):**
   Server komputasi AI menghasilkan panas ekstrem yang membutuhkan sistem pendingin evaporatif (*cooling towers*). Laporan keberlanjutan lingkungan Microsoft dan Google periode 2024 mencatat kenaikan konsumsi air tawar pendingin sebesar 17% hingga lebih dari 20% akibat intensifikasi infrastruktur AI. Pada tingkat operasional, setiap pelatihan model besar atau ribuan kueri inferensi inferensial setara dengan ratusan liter air tawar yang hilang ke atmosfer.
3. **Limbah Elektronik (*E-Waste*):**
   Siklus pergantian arsitektur GPU akselerator AI yang sangat cepat (siklus 2–3 tahun) mempercepat akumulasi limbah keras perangkat komputasi berperforma tinggi yang sulit didaur ulang secara optimal.

#### B. Solusi dan Langkah Mitigasi
1. **Optimalisasi Arsitektur Model (*Green AI*):**
   Mengurangi kebutuhan komputasi tanpa memangkas akurasi secara drastis melalui teknik kompresi model, seperti:
   - *Quantization* (konversi bobot ke INT8/INT4).
   - *Knowledge Distillation* (mentransfer kapabilitas ke model kecil).
   - Pemanfaatan metode adaptasi hemat parameter seperti LoRA/QLoRA untuk kebutuhan *fine-tuning*.
2. **Infrastruktur Bebas Karbon dan Pendinginan Cair Tertutup:**
   Operator pusat data beralih ke kontrak energi bebas karbon (PPA energi terbarukan/nuklir) serta mengganti pendingin evaporatif terbuka dengan teknologi *direct-to-chip liquid cooling* atau *immersion cooling* dengan siklus tertutup (*closed-loop*) agar tidak menghabiskan air bersih.
3. **Audit dan Transparansi Konsumsi Komputasi:**
   Mengintegrasikan alat pencatat emisi seperti *CodeCarbon* pada pipeline eksperimen untuk melacak konsumsi kWh dan emisi CO2e dari setiap sesi pelatihan model.

#### C. Pendapat Pribadi
Secara teknis, sering terjadi kecenderungan *over-engineering* di mana model *deep learning* skala besar digunakan untuk menyelesaikan persoalan yang sebenarnya cukup ditangani dengan model statistik atau machine learning konvensional (seperti LightGBM, Random Forest, atau regresi sederhana). Pendekatan rekayasa sistem yang bertanggung jawab harus memprioritaskan prinsip efisiensi komputasi (*fit-for-purpose*): gunakan arsitektur paling hemat energi yang mampu memenuhi ambang toleransi akurasi kebutuhan sistem.
